```python
## 🔢 Convert Numbers to Numeric
- Remove commas from numeric values
- Convert string columns into proper numeric datatypes
```

```python
df = df.replace(",", "", regex=True)
df = df.apply(pd.to_numeric, errors="ignore")
```

```python
## 🔄 Reshape Dataset (Wide → Long Format)
- Identify fiscal year (FY) columns
- Melt dataset to create a single 'year' column
- Store values in a unified 'value' column
```

```python
# Identify FY columns
fy_cols = [col for col in df.columns if "fy" in col.lower()]

# Melt only FY columns
df_melted = df.melt(id_vars=["data_provided_by_simfin"],
                    value_vars=fy_cols,
                    var_name="year",
                    value_name="value")
```

```python
## 📅 Clean and Convert Year Column
- Remove extra characters (FY, apostrophes, spaces)
- Keep only numeric year values
- Convert year column to integer datatype
```

```python
df_melted['year'] = (
    df_melted['year']
    .str.lower()
    .str.replace("fy", "")
    .str.replace("'", "")
    .str.strip()
)

# Convert only valid numeric values
df_melted = df_melted[df_melted['year'].str.isnumeric()]
df_melted['year'] = df_melted['year'].astype(int)
```
